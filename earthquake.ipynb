```python
import pandas as pd

url = ("https://earthquake.usgs.gov/fdsnws/event/1/query?format=csv"
       "&starttime=2020-01-01&endtime=2026-09-29&minmagnitude=3.5"
       "&minlatitude=32.5&maxlatitude=49.0&minlongitude=-125.5&maxlongitude=-114.1")

df = pd.read_csv(url)
states = df[df["place"].str.contains(r"\bCA\b|California|Oregon|Washington", na=False)]
print(len(states), "events")
```
